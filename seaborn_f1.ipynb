```python
import matplotlib.pyplot as plt

import seaborn as sns
import pandas as pd

df = pd.read_csv(
    "/Sample - Superstore.csv",
    encoding="latin1"
)

sns.histplot(data=df,x='Sales',bins=30,hue='Category',multiple='stack')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.kdeplot(
    data=df,
    x="Sales",
    fill=True,
    hue="Category"
)

plt.title("Distribution of Sales")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.rugplot(
    data=df,
    x="Sales",
    hue='Category'
)

plt.title("Sales Rug Plot")
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="Sales",
    fill=True
)

sns.rugplot(
    data=df,
    x="Sales",hue='Category'
)

plt.title("Sales Rug Plot")
plt.show()


plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.ecdfplot(
    data=df,
    x="Sales",
    hue="Category"
)

plt.title("ECDF of Sales")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(
    data=df,
    x="Category",
    y="Profit",
    hue="Region"
)

plt.title("Average Profit by Category")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=df,
    x="Category",
    hue="Region"
)

plt.title("Number of Orders by Category")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(
    data=df,
    x="Category",
    y="Profit",
    hue="Region",
    palette="Set1"

)
sns.stripplot(
    data=df,
    x="Category",
    y="Profit",
    hue="Region",
    palette="Set1",
    dodge=True,
    alpha=0.5
)
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
plt.title("Profit Distribution by Category")
plt.show()

```
